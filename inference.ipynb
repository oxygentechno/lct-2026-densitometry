```python
import sys
from pathlib import Path

from matplotlib import pyplot as plt

HERE = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "pipeline.py").is_file())
if str(HERE) not in sys.path:
    sys.path.insert(0, str(HERE))

from pipeline import DXAPipeline
```

```python
PATHS = [
    # "a.dcm",
]
if not PATHS:
    PATHS = sorted(HERE.rglob("*.dcm"))[:4]
PATHS = [str(p) for p in PATHS]
PATHS
```

```python
pipe = DXAPipeline()
frame, results = pipe.run_batch(PATHS)
frame
```

```python
for result in results:
    if result.visualization is None:
        print(result.path_to_study, result.processing_status)
        continue
    vis = pipe.draw_visualization(result)
    plt.figure(figsize=(12, 14))
    plt.imshow(vis)
    plt.axis("off")
    plt.title(f"{result.anatomical_region} | {result.path_to_study}")
    plt.tight_layout()
    plt.show()
```

```python
for result in results:
    if result.visualization is None:
        continue
    vis = pipe.draw_private_visualization(result)
    plt.figure(figsize=(6, 10))
    plt.imshow(vis)
    plt.axis("off")
    plt.title(result.path_to_study)
    plt.tight_layout()
    plt.show()
```
